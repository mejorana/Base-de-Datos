```python

import pymongo
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# 1. Conexion a MongoDB
client = pymongo.MongoClient("mongodb://localhost:27017/")
db = client["salud_mental_db"]
collection = db["intenciones"]

# 2. Cargar datos en Pandas
documentos = list(
    collection.find(
        {},
        {
            "_id": 0,
            "categoria": 1,
            "num_patrones": 1,
            "num_respuestas": 1
        }
    )
)

df = pd.DataFrame(documentos)

# 3. Regresion Lineal Simple
X = df[["num_patrones"]]
y = df["num_respuestas"]

modelo = LinearRegression()
modelo.fit(X, y)
y_pred = modelo.predict(X)

# 4. Metricas
m = modelo.coef_[0]
b = modelo.intercept_
r2 = r2_score(y, y_pred)

print(f"Ecuacion: Y = {m:.3f}X + {b:.3f}")
print(f"Coeficiente R2: {r2:.3f}")

# 5. Visualizacion
plt.figure(figsize=(8, 4))

sns.scatterplot(
    data=df,
    x="num_patrones",
    y="num_respuestas",
    color="blue",
    label="Datos"
)

plt.plot(
    X["num_patrones"],
    y_pred,
    color="red",
    label=f"Linea de Regresion (R2 = {r2:.2f})"
)

plt.title("Regresion Lineal: Patrones vs Respuestas")
plt.xlabel("Numero de Patrones (Mensajes de Usuario)")
plt.ylabel("Numero de Respuestas Configuradas")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
